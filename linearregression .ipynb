```python
import numpy as np

key = int(input("Enter key : "))

x = np.array([1, 2, 3, 4])
y = np.array([1.5,2.5,3.5,4.5])

x_mean = x.mean()
y_mean = y.mean()

xy_mean = x * y
x_y_mean = xy_mean.mean()

x_sqre = x**2
sqre_mean = x_sqre.mean()

xx_mean = x_mean**2

aa2 = x_y_mean - x_mean * y_mean
aa1 = sqre_mean - xx_mean
a1 = aa2 / aa1

a0 = y_mean - (a1 * x_mean)
y_ans = a0 + (a1 * key)


print(y_ans)
```
